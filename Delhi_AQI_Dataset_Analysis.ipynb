                                        Delhi Air Quality Index Prediction dataset Analysis 
This project on Jupyter Notebook  concists of the analysis of the Delhi AQI Weather Dataset. Here the prediction of AQI is made based on mainly 10 parrameters which are mentioned further in the project. 

As we all know that the Air Quality index is important to know the quality of air we are breathing in.  Basically, AQI is a system that translates the complex air pollution data into a single number, indicating the current and forecasted air health with higher numbers and redder coolours signalling worse pollution. 

And also every year Delhi has to face a very poor quality index due to pollution and stuble Burning. So for Delhi Rejuvenation, this project can be useful for predicting the Air Quality Index which can be used to predict the AQI of various laces of delhi and hence can help in taking further steps for the improvement of AQI. 

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
aqi_df = pd.read_csv("Delhi_AQI_Weather.csv") 
aqi_df 
```

                                            Now the processing of the data takes place

```python
# Removing of the  time column 
aqi_df= aqi_df.drop(["time_ist"], axis =1) 
aqi_df 
```

```python
# Removing of the date column  
aqi_df = aqi_df.drop(["date_ist"], axis = 1) 
aqi_df 
```

```python
print(aqi_df["location"].unique()) 
```

So basically, here 6 places have been taken into account and we have to cover the maximum part of Delhi which are as :

But still we have to remove the location  columns for easy analysis.  

```python
aqi_df = aqi_df.drop(['location'], axis = 1)  
aqi_df 
```

```python
print(aqi_df['condition_text'].unique())  
```

Above are the columns of the weather conditions which have to be remooved for the AQI Prediction. 

```python
aqi_df = aqi_df.drop(['condition_text'], axis = 1) 
aqi_df  
```

Now, at last the description has to be removed from tha data. Now we will get the perfect data required for the aqi prediction. 

```python
aqi_df2 = aqi_df.drop(['description'], axis = 1) 
aqi_df2 
```

This is the final data obtained for AQI Predition and now the AQI Index Prediction will take place based on 10 parameters : 
1. Latitude 
2. longitude 
3. Temperature in Celsius 
4. Humidity 
5. Atmospheric Pressure measured in Millibars 
6. Windspeed 
7. PM2.5 
8. PM10 
9. CO Level 
10. NO2 level 

Longitude and Latitude Positions affect the Air Quality Index by determining the local climate, sunlight intensity, wind pattersn and the proximity to the pollution sources. These coordinates allow for the identification of regional trends such as higher pollutant accumulation in the northern regions compared to southern regions. 

Temperature significantly impacts the Air Quality Index. It impacts the AQI by influencing the formation of the dispersion and trapping of pollutants.  High temperatures during the heatwaves increase the ozone and particle pollution. Conversely, due to cold temperatures can lead to temperature inversions that trap pollutants near the ground. 

Higher Humidity leads to higher AQI and poorer air quality by increasing the concentration of particulate matter. 
Hence humidity is directly proportional to AQI.  

Atmospheric Pressure affects the AQI by controlling the dispersion of pollutants. High pressure systems create the stable, stagnant air that traps the pollutant near the ground leading to higher AQI.  Also the low pressure systems  provide the air circulation, wind, rain which disperse the pollutants and improve the  air quality. 

Windspeed affects the AQI of the environment as they disperse the pollutants.  High windspeeds disperse the pollutants and help in improving the AQi whereas the low wind speeds make the pollutants stagnant and it worsens the AQI. 

              Now, we do the analysis of the data aqi_df2 here by applying the machine learning algorithm. 

Dividing the dataset into the dependent and independent features. 

Independent features are those, based on which the prediction has to be made whereas the dependent feature is the prediction outcome itself. 

```python
x = aqi_df2.drop(['aqi_index'], axis = 1) 
x 
```

```python
y = aqi_df2.filter(['aqi_index'], axis = 1)
y 
```

                 Now, we have to rain and test the model by using the train_test-split model 

```python
from sklearn.model_selection import train_test_split 
x_train, x_test, y_train, y_test = train_test_split(x,y,  test_size = 0.20, random_state = 0) 
```

```python
x_train # Data on whic  the model is trained 
```

```python
x_test # Data on which model is trained 
```

```python
y_train # data on which the model is trained 
```

```python
y_test # data on which model is tested 
```

Now, we have to scale the data using the Standard Scaling Process. 

```python
from sklearn.preprocessing import StandardScaler 
sc = StandardScaler() 

x_train = sc.fit_transform(x_train)  
x_test = sc.transform(x_test)
```

```python
# Training the Random Forest Regressor Model  
from sklearn.ensemble import RandomForestRegressor 

model = RandomForestRegressor(n_estimators = 500, random_state = 42) 

regressor = model.fit(x_train, y_train) 
regressor 
```

```python
y_pred = regressor.predict(x_test) 
y_pred 
```

Now, we have to check the accuracy of the model by various parameters. These parameters will help us to understand that whether a model is working perfectly or not. 

1. Mean Absolute error : Mean Absolute Error is basically calculated by taking the average of the absolute error obtained by subtracting the real values from the predcted values. 

2. Mean Squared Error : Mean Squared Error is somewhat similar to the Mean Absolute error. But here, what happens is the difference of the predicted and the average values is taken and the square is taken. 

3. Root Mean Square Error : It is the square root of the Mean Square Error.   

4. R2_Score : R2-Score indicates the proportion of variance of  in the dependent variables that's predictable from the independent variables. The closer the score to 1, more perfect is the model. 

```python
from sklearn.metrics import mean_absolute_error, r2_score, mean_squared_error  

print("Mean Absolute Error: ", mean_absolute_error(y_test, y_pred)) 

print("Mean Squared Error: ", mean_squared_error(y_test, y_pred)) 

print("Root Mean Squared Error: ", np.sqrt(mean_squared_error(y_test, y_pred))) 

print("R2 Score : ", r2_score(y_test, y_pred)) # R@ score is also like the accuracy 
```

The R2 Score value is basically referred tthe model's accuracy. Hence here the R2 score is around 0.93 which is around 93 percent.
